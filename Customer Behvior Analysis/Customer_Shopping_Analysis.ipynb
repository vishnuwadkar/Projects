```python
#choose median over mean to fill null values, because mean can get affected by outliers
#we will find the median value of each category, and impute those values to that category in order to eliminate bias
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

### Thus we cleaned the data by imputing the null values, in a more smarter way

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
df = df.rename(columns = {'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

# Feature Engeineering
### Grouping customers into four groups

```python
labels = ['Young Adult', 'Adult','Middle Aged', 'Senior']
df['age_group'] = pd.qcut(df['age'],q=4,labels=labels)
```

```python
df[['age','age_group']].head(5)
```

### Creating another column : purchase_frequency_days

```python
frequency_mapping = {
    'Fortnightly':14,
    'Weekly':7,
    'Monthly':30,
    'Quaterly':90,
    'Bi-weekly':14,
    'Annually':365,
    'Every 3 Months' : 90
  }

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['frequency_of_purchases','purchase_frequency_days']].head()
```

```python
import urllib.parse
from sqlalchemy import create_engine, text

username = "postgres"
password = "Vishnu1210@"
host = "127.0.0.1"
port = "5432"
database = "customer_behaviour"

escaped_password = urllib.parse.quote_plus(password)

# 1. Connect to the default 'postgres' database first to create the target database if it doesn't exist
temp_engine = create_engine(f"postgresql+psycopg2://{username}:{escaped_password}@{host}:{port}/postgres", isolation_level="AUTOCOMMIT")

with temp_engine.connect() as conn:
    # Check if database exists
    result = conn.execute(text(f"SELECT 1 FROM pg_database WHERE datname='{database}'"))
    exists = result.scalar()
    if not exists:
        conn.execute(text(f"CREATE DATABASE {database}"))
        print(f"Database '{database}' created successfully!")
    else:
        print(f"Database '{database}' already exists.")

temp_engine.dispose()

# 2. Now connect to the newly created 'customer_behaviour' database and write the dataframe
engine = create_engine(f"postgresql+psycopg2://{username}:{escaped_password}@{host}:{port}/{database}")

table_name = "customer_table"
df.to_sql(table_name, engine, if_exists='replace', index=False)

print(f"Data successfully loaded into table '{table_name}' in database '{database}'!")
```
